```python
SEED = 42
EPOCHS = 50
BATCH = 16
IMGSZ = 640
OPTIMIZER = "Adam"
LR0 = 0.001            # paper doesn't give lr; same assumption as experiment A
MODELS = ["yolov8m.pt", "yolov8s.pt"]

REPO = "https://github.com/YOUANNA13/-1_proj.git"
CHV_DRIVE_ID = "1fdGn67W0B7ShpBDbbQpUF0ScPQa4DR0a"
EGY_DRIVE_ID = "1Y30bp2I8c5RGSxtRb0Xa6t5RjGmzxx8n"
NAMES = ["person", "vest", "blue_helmet", "red_helmet", "white_helmet", "yellow_helmet"]  # D-005
ROTATE_CCW = {"IMG (164)"}   # sideways photo with no EXIF tag; its labels are already upright
```

```python
import subprocess, zipfile, shutil
from PIL import Image, ImageOps

# our repo: splits + reviewed Egyptian labels
shutil.rmtree("repo", ignore_errors=True)
subprocess.run(["git", "clone", "-q", "--depth", "1", REPO, "repo"], check=True)

# raw datasets
if not Path("chv.zip").exists():
    subprocess.run(["gdown", "--fuzzy", f"https://drive.google.com/file/d/{CHV_DRIVE_ID}/view", "-O", "chv.zip"], check=True)
if not Path("egy.rar").exists():
    subprocess.run(["gdown", "--fuzzy", f"https://drive.google.com/file/d/{EGY_DRIVE_ID}/view", "-O", "egy.rar"], check=True)
with zipfile.ZipFile("chv.zip") as z:
    z.extractall("chv")
subprocess.run(["unrar", "x", "-o+", "-idq", "egy.rar", "egy_raw/"], check=True)

chv_imgs = {p.name: p for p in Path("chv").rglob("*.jpg") if "__MACOSX" not in p.parts}
chv_anns = {p.stem: p for p in Path("chv").rglob("ppe_*.txt") if "__MACOSX" not in p.parts}
egy_raw = {p.stem: p for p in Path("egy_raw").rglob("*") if p.suffix.lower() in (".jpg", ".jpeg", ".png")}

D = W / "datasets"
shutil.rmtree(D, ignore_errors=True)
for part, folder in {"train": "train", "valid": "val", "test": "test"}.items():
    for ds in ("chv", "egy"):
        (D / ds / "images" / folder).mkdir(parents=True, exist_ok=True)
        (D / ds / "labels" / folder).mkdir(parents=True, exist_ok=True)

    # CHV: official split, labels as-is
    chv_list = [l for l in Path(f"repo/splits/chv_{part}.txt").read_text().split() if l.strip()]
    for rel in chv_list:
        name = Path(rel).name
        shutil.copy(chv_imgs[name], D / "chv/images" / folder / name)
        shutil.copy(chv_anns[Path(name).stem], D / "chv/labels" / folder / (Path(name).stem + ".txt"))

    # Egyptian: our split, upright images, reviewed labels
    egy_list = [Path(l).stem for l in Path(f"repo/splits/egy_{part}.txt").read_text().split("\n") if l.strip()]
    for stem in egy_list:
        with Image.open(egy_raw[stem]) as im:
            im = ImageOps.exif_transpose(im).convert("RGB")
            if stem in ROTATE_CCW:
                im = im.rotate(90, expand=True)
            im.save(D / "egy/images" / folder / (stem + ".jpg"), quality=95)
        shutil.copy(f"repo/annotations/egy/{stem}.txt", D / "egy/labels" / folder / (stem + ".txt"))

    print(f"{folder}: chv {len(chv_list)}  egy {len(egy_list)}")   # expect 1064/106, 133/13, 133/14

names_yaml = "names:\n" + "".join(f"  {i}: {n}\n" for i, n in enumerate(NAMES))
Path("chv_egy.yaml").write_text(
    f"path: {D}\ntrain:\n  - chv/images/train\n  - egy/images/train\n"
    f"val:\n  - chv/images/val\n  - egy/images/val\ntest:\n  - chv/images/test\n  - egy/images/test\n" + names_yaml)
for ds in ("chv", "egy"):
    Path(f"{ds}.yaml").write_text(f"path: {D/ds}\ntrain: images/train\nval: images/val\ntest: images/test\n" + names_yaml)
```

```python
from ultralytics import YOLO

best_weights = {}
for m in MODELS:
    name = f"repro_{Path(m).stem}_chv_egy"
    model = YOLO(m)                # COCO-pretrained, like the paper
    model.train(data="chv_egy.yaml", epochs=EPOCHS, batch=BATCH, imgsz=IMGSZ,
                optimizer=OPTIMIZER, lr0=LR0, seed=SEED, deterministic=True,
                project=str(SAVE), name=name, exist_ok=True)
    best_weights[name] = Path(model.trainer.best)
    print(name, "->", best_weights[name])
```

```python
import time, csv, datetime, torch
import pandas as pd
from ultralytics.utils.torch_utils import get_flops

gpu = torch.cuda.get_device_name(0)
EVALS = {"chv_test": "chv.yaml", "egy_test": "egy.yaml", "all_test": "chv_egy.yaml"}

# experiment A weights (trained on CHV only) - evaluated on the same test sets for the A vs B comparison
runs = dict(best_weights)
for m in MODELS:
    a = SAVE / f"repro_{Path(m).stem}_chv" / "weights" / "best.pt"
    if a.exists():
        runs[f"repro_{Path(m).stem}_chv"] = a

rows, per_class = [], []
for name, w in runs.items():
    model = YOLO(str(w))
    params = sum(p.numel() for p in model.model.parameters())
    gflops = get_flops(model.model, IMGSZ)

    # FPS: batch 1, after warm-up, full pipeline, on the CHV test images
    test_imgs = sorted((D / "chv/images/test").glob("*.jpg"))
    for p in test_imgs[:10]:
        model.predict(str(p), imgsz=IMGSZ, verbose=False)
    t = time.perf_counter()
    for p in test_imgs:
        model.predict(str(p), imgsz=IMGSZ, verbose=False)
    fps = len(test_imgs) / (time.perf_counter() - t)

    for split_name, yaml in EVALS.items():
        m = model.val(data=yaml, split="test", imgsz=IMGSZ, batch=16,
                      project=str(SAVE), name=f"{name}_{split_name}", exist_ok=True)
        P, R = m.box.mp, m.box.mr
        rows.append(dict(run_id=name, trained_on="chv+egy" if "chv_egy" in name else "chv",
                         eval_split=split_name, date=datetime.date.today().isoformat(),
                         imgsz=IMGSZ, epochs=EPOCHS, batch=BATCH, optimizer=OPTIMIZER, lr0=LR0, seed=SEED,
                         hardware=gpu, precision=round(P, 3), recall=round(R, 3),
                         f1=round(2 * P * R / (P + R + 1e-9), 3),
                         map50=round(m.box.map50, 3), map50_95=round(m.box.map, 3),
                         fps=round(fps, 1), params_m=round(params / 1e6, 2), gflops=round(gflops, 2)))
        for i, c in enumerate(m.box.ap_class_index):     # classes that appear in this test set
            p, r = m.box.p[i], m.box.r[i]
            per_class.append(dict(run_id=name, eval_split=split_name, cls=NAMES[int(c)],
                                  P=round(p, 3), R=round(r, 3), F1=round(2 * p * r / (p + r + 1e-9), 3),
                                  AP50=round(m.box.ap50[i], 3), AP50_95=round(m.box.ap[i], 3)))

for fname, data in [("results_B_overall.csv", rows), ("results_B_per_class.csv", per_class)]:
    with open(SAVE / fname, "w", newline="") as f:
        wr = csv.DictWriter(f, fieldnames=data[0].keys()); wr.writeheader(); wr.writerows(data)

display(pd.DataFrame(rows)); display(pd.DataFrame(per_class))
```
